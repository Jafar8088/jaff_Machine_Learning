```python
# Lets our model using sklearn
# 1st step is use train_test_split to split your dataset into training and testing
# df dataframe has target column lets remove the target column in df dataset
# Look at this dataframe first 4 columns are tghe features that you want to use for training your model.
# 2nd two columns are the target columns...ok..
# I want to drop target and flower_name.

from sklearn.model_selection import train_test_split

X = df.drop(['target','flower_name'],axis = 'columns')
X.head()
```

```python
y = df.target
```

```python
# Lets use train_test_split x
# and y is test size Lets say I want to use 20% of sample as a test and 80% as a training
# So x train the output of this train_test_split method is x, train x 
# test y train 
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2)
```

```python
# look the length of your X (20% of test, and 80% of train)
print('The length of 80% of X_train_data is :',len(X_train))
print('The length of 20% of X_test data is :',len(X_test))
```

```python
# Create SVM Classifier model
from sklearn.svm import SVC
model = SVC(kernel='linear')
```

```python
# As usual you can call fit() method ti train your model
# Here iam going to use X_train and y_train
model.fit(X_train,y_train)
# C = 1.0 is a regulaization parameter, you have kernal and gamma also
```

```python
# Look at the accuracy of our model using score() method
# now this time Iam going to use X_test and y_test
model.score(X_test,y_test)
```

```python
# You can seperate the paramater for example my default c = 1.0
# Iam modify the C parameter leys say you do C = 10, model = SVC(C = 10) in the above 
# You can use gamma also model = SVC(gamma = 100) in the above, it make model performance is worse 
# that is 0.56666 
# yoy can also use kernal model = SVC(kernal = 'linear') in the above,By default ketrnal was RBF

print(model.score(X_test,y_test))
```
