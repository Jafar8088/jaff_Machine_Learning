# Step 6 : Outliers Treatments

```python
# Decide whether to do Outliers treament or not , if we do how ?
```

We need to figure out this first which are the column are the numerical contineuous numeracal data for that column we will do the outliers treatment.

```python
# For Outlier treatment first we need to get the "Upper Whisker" and "Lower Whisker first"
# that have to define one "whisker_function" and give input as columns i.e "col"
# In order to find the upper wisker and lower wisker we need to do first get the "quartile"
# q1 and q3 are 25th percentile and 75th percentile of dat is equal np.percentile()
# Find the interquarter range (IQR)
# after finding IQR i will find lower wisker (lw), then upper wisker (uw)
# after calculating, return lower wisker first , upper wisker
def wisker(col):
    q1,q3 = np.percentile(col,[25,75])
    iqr = q3 - q1
    lw = q1 - 1.5 * iqr 
    uw = q3 + 1.5 * iqr
    return lw,uw
```

```python
# lets check the column information for each
df.columns
```

```python
wisker(df['GDP'])
# it will give the "lower wisker" and "upper wisker" of GDP column
```

```python
# fill the upper wisker and lower wisker to these 4 column's I selected

for i in ['GDP','Total expenditure','thinness  1-19 years','thinness 5-9 years']:
    lw,uw = wisker(df[i])
    df[i] = np.where(df[i] < lw,lw,df[i]) # check df[i] < lw fill with lower wisker otherwise keep it as it is
    df[i] = np.where(df[i] > uw,uw,df[i]) # check df[i] > uw fill upper wisker otherwise keep it as it is
```
