# Important date-format list

| Format | Meaning          | Example     |
| ------ | ---------------- | ----------- |
| `%d`   | Day (01–31)      | `25`        |
| `%m`   | Month (01–12)    | `09`        |
| `%Y`   | Year, 4 digits   | `2026`      |
| `%y`   | Year, 2 digits   | `26`        |
| `%B`   | Full month name  | `September` |
| `%b`   | Short month name | `Sep`       |
| `%A`   | Full day name    | `Friday`    |
| `%a`   | Short day name   | `Fri`       |


| Format | Meaning      | Example |
| ------ | ------------ | ------- |
| `%H`   | Hour (00–23) | `14`    |
| `%I`   | Hour (01–12) | `02`    |
| `%M`   | Minute       | `30`    |
| `%S`   | Second       | `45`    |
| `%p`   | AM/PM        | `PM`    |

```python
# 10. Concatinate and Merge Dataframes 
# Concatinate means 'Copy and Past'
# Merge means 'Joins'
```

```python
# Create two dataframe
df1a = pd.DataFrame({'ID' : [1,2,3],'Name' : ['A','B','C']})
df1a
```

```python
df1b = pd.DataFrame({'ID' : [1,2,2,4], 'Score' : [88,96,77,79]})
df1b
```

```python
# concatinate two dataframe
pd.concat([df1a,df1b],axis = 0) # axis = 0 means vertical/row level/ top on top
```

```python
pd.concat([df1a,df1b],axis = 1) # Horizontal/ Column Level / Side on Side
```

```python
# Merge = JOIN

pd.merge(df1a,df1b, how = 'inner', on = 'ID')
```

```python
pd.merge(df1a,df1b, how = 'inner', left_on = 'ID', right_on = 'ID')
```

```python
# 11. AI and Chatgpt :
# Generate one prompt 
# Prompt is For the given dataframe, write a pandas script to filter employees whose joining date 
# is in the month of January and whose salary is greater then 70,000
```
